# 🩺 Esplorazione Dataset VitalDB — Layer Bronze

Questo notebook permette di esplorare interattivamente il dataset **VitalDB**:
1. Caricamento dei dati clinici (demografia, tipo intervento, durata)
2. Ricerca dei casi chirurgici contenenti le tracce biometriche di nostro interesse (`HR`, `SpO2`, `NIBP_SBP`, `NIBP_DBP`, `NIBP_MBP`)
3. Caricamento e visualizzazione grafica di un caso reale

```python
import sys
from pathlib import Path

# Aggiunge la root del progetto al sys.path
ROOT_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.append(str(ROOT_DIR))

import vitaldb
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.config import VITAL_TRACKS, SAMPLING_INTERVAL

print("✓ Librerie importate con successo.")
print("Tracche di interesse:", VITAL_TRACKS)
```

## 1. Caricamento Dati Clinici
VitalDB mette a disposizione 74 parametri clinici e di laboratorio per i ~6.388 casi chirurgici.

```python
df_clinical = vitaldb.load_clinical_data()
print(f"Totale casi clinici trovati: {len(df_clinical)}")
df_clinical.head(10)
```

## 2. Ricerca dei Casi con le Tracce Biometriche Scelte
Troviamo i `case_id` che contengono le nostre tracce di interesse.

```python
# Cerca i casi che contengono Frequenza Cardiaca (HR) e Saturazione (SpO2)
cases_hr_spo2 = vitaldb.find_cases(["Solar8000/HR", "Solar8000/PLETH_SPO2"])
print(f"Casi che contengono HR e SpO2: {len(cases_hr_spo2)}")

# Primi 20 ID casi disponibili
print("Primi 20 casi:", cases_hr_spo2[:20])
```

## 3. Caricamento e Visualizzazione di un Caso Reale
Carichiamo i dati time-series per il primo caso disponibile a intervallo di 1 secondo.

```python
sample_case_id = cases_hr_spo2[0]
print(f"Caricamento caso #{sample_case_id}...")

# Carica il caso con le tracce specificate
data = vitaldb.load_case(sample_case_id, VITAL_TRACKS, interval=SAMPLING_INTERVAL)
df_sample = pd.DataFrame(data, columns=VITAL_TRACKS)

# Aggiunge colonna tempo in secondi e minuti
df_sample['Time_sec'] = np.arange(len(df_sample)) * SAMPLING_INTERVAL
df_sample['Time_min'] = df_sample['Time_sec'] / 60.0

print(f"Shape del DataFrame: {df_sample.shape}")
df_sample.head()
```

## 4. Visualizzazione Grafica delle Serie Temporali

```python
plt.figure(figsize=(14, 10))

# Heart Rate (HR)
plt.subplot(3, 1, 1)
plt.plot(df_sample['Time_min'], df_sample['Solar8000/HR'], color='red', label='HR (bpm)')
plt.title(f'Caso VitalDB #{sample_case_id} — Heart Rate (HR)')
plt.xlabel('Tempo (minuti)')
plt.ylabel('bpm')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()

# SpO2
plt.subplot(3, 1, 2)
plt.plot(df_sample['Time_min'], df_sample['Solar8000/PLETH_SPO2'], color='blue', label='SpO2 (%)')
plt.title('Saturazione Ossigeno (SpO2)')
plt.xlabel('Tempo (minuti)')
plt.ylabel('%')
plt.ylim(50, 105)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()

# Pressione Sanguigna (NIBP)
plt.subplot(3, 1, 3)
if 'Solar8000/NIBP_SBP' in df_sample:
    plt.plot(df_sample['Time_min'], df_sample['Solar8000/NIBP_SBP'], color='orange', label='Pressione Sistolica (SBP)')
if 'Solar8000/NIBP_DBP' in df_sample:
    plt.plot(df_sample['Time_min'], df_sample['Solar8000/NIBP_DBP'], color='green', label='Pressione Diastolica (DBP)')
plt.title('Pressione Sanguigna (NIBP)')
plt.xlabel('Tempo (minuti)')
plt.ylabel('mmHg')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()

plt.tight_layout()
plt.show()
```
